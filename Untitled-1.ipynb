```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB, GaussianNB, MultinomialNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
Dataset = pd.read_csv("weather_forecast_data.csv")
```

```python
Dataset.head() 
```

```python
Dataset.tail() 
```

```python
print('rows: ', Dataset.shape[0]) 
print('columns: ', Dataset.shape[1]) 
```

```python
Dataset.isnull().sum()
```

```python
Dataset = Dataset.fillna(Dataset.mean) 
Dataset = Dataset.fillna(Dataset.median)
Dataset = Dataset.fillna(Dataset.mode)
```

```python
Dataset.dtypes
```

```python
Dataset["Temperature"].unique()
```

```python
Dataset["Temperature"] = Dataset["Temperature"].astype(np.int64)
Dataset["Humidity"] = Dataset["Humidity"].astype(np.int64)
Dataset["Wind_Speed"] = Dataset["Wind_Speed"].astype(np.int64)
Dataset["Cloud_Cover"] = Dataset["Cloud_Cover"].astype(np.int64)
Dataset["Pressure"] = Dataset["Pressure"].astype(np.int64)
```

```python
x= Dataset.iloc[:, 0:-1] 
x.shape 
y = Dataset.iloc[:,-1] 
y.shape 
```

```python
cat_columns = x.select_dtypes(['object']).columns 
x[cat_columns] = x[cat_columns].apply(lambda x: pd.factorize(x)[0]) 
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x,y, test_size=.3, shuffle=False)
```

```python
results = {}
BernNB = BernoulliNB()
BernNB.fit(X_train, Y_train)
Y_bpred = BernNB.predict(X_test)
results['Bernoulli'] = {
    'Accuracy': accuracy_score(Y_test, Y_bpred),
    'Precision': precision_score(Y_test, Y_bpred, average='weighted', labels=np.unique(Y_bpred)),
    'Recall': recall_score(Y_test, Y_bpred, average='weighted', labels=np.unique(Y_bpred)),
    'F1': f1_score(Y_test, Y_bpred, average='weighted', labels=np.unique(Y_bpred))
}
RF = RandomForestClassifier()
RF.fit(X_train, Y_train)
Y_rpred = RF.predict(X_test)
results['Random Forest'] = {
    'Accuracy': accuracy_score(Y_test, Y_rpred),
    'Precision': precision_score(Y_test, Y_rpred, average='weighted', labels=np.unique(Y_rpred)),
    'Recall': recall_score(Y_test, Y_rpred, average='weighted', labels=np.unique(Y_rpred)),
    'F1': f1_score(Y_test, Y_rpred, average='weighted', labels=np.unique(Y_rpred))
}
GausNB = GaussianNB()
GausNB.fit(X_train, Y_train)
Y_gpred = GausNB.predict(X_test)
results['Gaussian'] = {
    'Accuracy': accuracy_score(Y_test, Y_gpred),
    'Precision': precision_score(Y_test, Y_gpred, average='weighted', labels=np.unique(Y_gpred)),
    'Recall': recall_score(Y_test, Y_gpred, average='weighted', labels=np.unique(Y_gpred)),
    'F1': f1_score(Y_test, Y_gpred, average='weighted', labels=np.unique(Y_gpred))
}
Dtree = DecisionTreeClassifier()
Dtree.fit(X_train, Y_train)
Y_dpred = Dtree.predict(X_test)
results['Decision Tree'] = {
    'Accuracy': accuracy_score(Y_test, Y_dpred),
    'Precision': precision_score(Y_test, Y_dpred, average='weighted', labels=np.unique(Y_dpred)),
    'Recall': recall_score(Y_test, Y_dpred, average='weighted', labels=np.unique(Y_dpred)),
    'F1': f1_score(Y_test, Y_dpred, average='weighted', labels=np.unique(Y_dpred))
}
MultiNB = MultinomialNB()
MultiNB.fit(X_train, Y_train)
Y_mpred = MultiNB.predict(X_test)
results['Multinomial'] = {
    'Accuracy': accuracy_score(Y_test, Y_mpred),
    'Precision': precision_score(Y_test, Y_mpred, average='weighted', labels=np.unique(Y_mpred)),
    'Recall': recall_score(Y_test, Y_mpred, average='weighted', labels=np.unique(Y_mpred)),
    'F1': f1_score(Y_test, Y_mpred, average='weighted', labels=np.unique(Y_mpred))
}
KNN = KNeighborsClassifier()
KNN.fit(X_train, Y_train)
Y_kpred = KNN.predict(X_test)
results['KNeighbors'] = {
    'Accuracy': accuracy_score(Y_test, Y_kpred),
    'Precision': precision_score(Y_test, Y_kpred, average='weighted', labels=np.unique(Y_kpred)),
    'Recall': recall_score(Y_test, Y_kpred, average='weighted', labels=np.unique(Y_kpred)),
    'F1': f1_score(Y_test, Y_kpred, average='weighted', labels=np.unique(Y_kpred))
}
for clf in results:
    print(f"\n{clf} Scores:")
    for metric in results[clf]:
        print(f"{metric}: {results[clf][metric]:.4f}")
plt.figure(figsize=(16, 10))
metrics_list = ['Accuracy', 'Precision', 'Recall', 'F1']
for metric_name in metrics_list:
    y_vals = [results[clf][metric_name] for clf in results]
    plt.plot(list(results.keys()), y_vals, marker='o', label=metric_name)
plt.title("Scores of Applied Classifiers")
plt.legend()
plt.show()
plt.figure(figsize=(16, 10))
plt.bar(range(len(results)), [results[clf]['F1'] for clf in results],
        tick_label=list(results.keys()),
        color=['#08737f', '#00898a', '#089f8f', '#39b48e', '#64c987', '#92dc7e'])
plt.xlabel('Classifiers')
plt.ylabel('F1 Scores')
plt.title('F1 Scores of Applied Classifiers')
plt.show()
```
